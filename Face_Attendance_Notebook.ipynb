```python
#Helper Function 

import os
import cv2
import csv
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from tqdm import tqdm
from datetime import datetime
from collections import defaultdict
from sklearn.metrics import (
    accuracy_score,
    precision_recall_fscore_support,
    classification_report,
    confusion_matrix,
    roc_curve,
    auc
)
from sklearn.preprocessing import label_binarize

warnings.filterwarnings("ignore")

def l2_normalize(x, eps=1e-10):
    return x / (np.linalg.norm(x) + eps)

def cosine_similarity(a, b):
    a = l2_normalize(a)
    b = l2_normalize(b)
    return float(np.dot(a, b))

def read_image(path):
    img = cv2.imread(str(path))
    if img is None:
        raise ValueError(f"Could not read image: {path}")
    return img

def list_images(folder):
    exts = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}
    return [p for p in Path(folder).rglob("*") if p.suffix.lower() in exts]

def detect_faces_and_embeddings(img_bgr):
    faces = app.get(img_bgr)
    valid_faces = []
    for face in faces:
        x1, y1, x2, y2 = map(int, face.bbox)
        w = x2 - x1
        h = y2 - y1
        if min(w, h) >= MIN_FACE_SIZE:
            valid_faces.append(face)
    return valid_faces

def extract_best_face_embedding(img_bgr):
    faces = detect_faces_and_embeddings(img_bgr)
    if len(faces) == 0:
        return None, None

    best_face = sorted(faces, key=lambda f: float(f.det_score), reverse=True)[0]
    emb = l2_normalize(best_face.embedding.astype(np.float32))
    bbox = tuple(map(int, best_face.bbox))
    return emb, bbox

def plot_confusion_matrix(cm, labels, figsize=(10, 8), title="Confusion Matrix"):
    plt.figure(figsize=figsize)
    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues",
                xticklabels=labels, yticklabels=labels)
    plt.title(title)
    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.tight_layout()
    plt.show()
```

```python
#Attendance helpers

def create_attendance_file_if_missing(csv_path):
    if not Path(csv_path).exists():
        with open(csv_path, "w", newline="", encoding="utf-8") as f:
            writer = csv.writer(f)
            writer.writerow(["Name", "Date", "Time", "Timestamp", "Camera_ID", "Confidence"])

def save_attendance_to_excel(csv_path, excel_path):
    if Path(csv_path).exists():
        df = pd.read_csv(csv_path)
        df.to_excel(excel_path, index=False)

def append_attendance(csv_path, name, camera_id, confidence, excel_path=None):
    now = datetime.now()
    date_str = now.strftime("%Y-%m-%d")
    time_str = now.strftime("%H:%M:%S")
    timestamp = now.isoformat(timespec="seconds")

    with open(csv_path, "a", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow([name, date_str, time_str, timestamp, camera_id, round(confidence, 4)])

    if excel_path is not None:
        save_attendance_to_excel(csv_path, excel_path)
```

```python
#Build Gallery function

def build_face_gallery(gallery_dir):
    gallery_dir = Path(gallery_dir)
    if not gallery_dir.exists():
        raise ValueError(f"Gallery directory not found: {gallery_dir}")

    person_to_embeddings = defaultdict(list)
    skipped = []

    persons = [d for d in gallery_dir.iterdir() if d.is_dir()]
    print(f"Found {len(persons)} enrolled identities.")

    for person_dir in tqdm(persons, desc="Building gallery"):
        person_name = person_dir.name
        img_paths = list_images(person_dir)

        for img_path in img_paths:
            try:
                img = read_image(img_path)
                emb, bbox = extract_best_face_embedding(img)
                if emb is not None:
                    person_to_embeddings[person_name].append(emb)
                else:
                    skipped.append(str(img_path))
            except Exception:
                skipped.append(str(img_path))

    gallery_centroids = {}
    gallery_counts = {}

    for person, embs in person_to_embeddings.items():
        if len(embs) == 0:
            continue
        centroid = np.mean(np.vstack(embs), axis=0)
        centroid = l2_normalize(centroid)
        gallery_centroids[person] = centroid
        gallery_counts[person] = len(embs)

    gallery_meta = pd.DataFrame({
        "person": list(gallery_counts.keys()),
        "num_images_used": list(gallery_counts.values())
    }).sort_values("person").reset_index(drop=True)

    print(f"Gallery built for {len(gallery_centroids)} people.")
    print(f"Skipped images: {len(skipped)}")

    return gallery_centroids, person_to_embeddings, gallery_meta, skipped
```

```python
#Prediction Function

def predict_identity_from_embedding(embedding, gallery_centroids, threshold=0.45):
    scores = {}
    for person, centroid in gallery_centroids.items():
        scores[person] = cosine_similarity(embedding, centroid)

    best_person = max(scores, key=scores.get)
    best_score = scores[best_person]

    if best_score < threshold:
        return "Unknown", best_score, scores

    return best_person, best_score, scores
```

```python
#Evaluating Test Set

def evaluate_test_set(test_dir, gallery_centroids, threshold=0.45, save_csv_path=None):
    test_dir = Path(test_dir)
    if not test_dir.exists():
        raise ValueError(f"Test directory not found: {test_dir}")

    rows = []
    skipped = []

    test_person_dirs = [d for d in test_dir.iterdir() if d.is_dir()]
    known_classes = sorted(list(gallery_centroids.keys()))

    for person_dir in tqdm(test_person_dirs, desc="Evaluating test set"):
        true_label = person_dir.name
        img_paths = list_images(person_dir)

        for img_path in img_paths:
            try:
                img = read_image(img_path)
                emb, bbox = extract_best_face_embedding(img)
                if emb is None:
                    skipped.append(str(img_path))
                    continue

                pred_label, score, all_scores = predict_identity_from_embedding(
                    emb, gallery_centroids, threshold=threshold
                )

                row = {
                    "image_path": str(img_path),
                    "true_label": true_label,
                    "pred_label": pred_label,
                    "score": score
                }

                for cls in known_classes:
                    row[f"score_{cls}"] = all_scores.get(cls, np.nan)

                rows.append(row)

            except Exception:
                skipped.append(str(img_path))

    df = pd.DataFrame(rows)

    if df.empty:
        raise ValueError("No predictions produced. Check your test folder.")

    if save_csv_path is not None:
        df.to_csv(save_csv_path, index=False)

    y_true = df["true_label"].values
    y_pred = df["pred_label"].values

    summary = {}
    summary["accuracy_overall"] = accuracy_score(y_true, y_pred)

    prec, rec, f1, _ = precision_recall_fscore_support(
        y_true, y_pred, average="weighted", zero_division=0
    )

    summary["precision_weighted"] = prec
    summary["recall_weighted"] = rec
    summary["f1_weighted"] = f1
    summary["num_samples"] = len(df)
    summary["num_skipped"] = len(skipped)
    summary["num_unknown_predictions"] = int((df["pred_label"] == "Unknown").sum())

    return df, summary, skipped
```

```python
#Running the Evaluation

eval_df, summary, skipped_test = evaluate_test_set(
    TEST_DIR,
    gallery_centroids,
    threshold=SIMILARITY_THRESHOLD,
    save_csv_path=EVAL_CSV
)

print("Evaluation Summary")
for k, v in summary.items():
    print(f"{k}: {v}")
```

```python
import random

def evaluate_fairface_subset(df, base_dir, gallery_centroids, sample_size=200):
    df_sample = df.sample(n=min(sample_size, len(df)), random_state=42)

    results = []

    for _, row in tqdm(df_sample.iterrows(), total=len(df_sample), desc="FairFace Eval"):
        img_rel_path = row["file"]
        img_path = base_dir / img_rel_path

        if not img_path.exists():
            continue

        try:
            img = read_image(img_path)
            emb, _ = extract_best_face_embedding(img)

            if emb is None:
                continue

            pred_label, score, _ = predict_identity_from_embedding(
                emb, gallery_centroids, threshold=SIMILARITY_THRESHOLD
            )

            results.append({
                "race": row["race"],
                "gender": row["gender"],
                "age": row["age"],
                "pred_label": pred_label,
                "confidence": score
            })

        except:
            continue

    return pd.DataFrame(results)
```

```python
def extract_embedding_from_cropped_face(img_bgr):
    """
    FairFace images are already cropped face images.
    So we bypass face detection and send the whole image to the recognition model.
    """
    face_crop = cv2.resize(img_bgr, (112, 112))
    emb = rec_model.get_feat(face_crop)

    if emb is None:
        return None

    emb = np.array(emb).flatten().astype(np.float32)
    emb = l2_normalize(emb)
    return emb
```

```python
def evaluate_fairface_subset(df, base_dir, gallery_centroids, sample_size=200):
    df_sample = df.sample(n=min(sample_size, len(df)), random_state=42)

    results = []

    for _, row in tqdm(df_sample.iterrows(), total=len(df_sample), desc="FairFace Eval"):
        img_rel_path = row["file"]
        img_path = base_dir / img_rel_path

        if not img_path.exists():
            continue

        try:
            img = read_image(img_path)

            # Use direct embedding extraction for cropped FairFace images
            emb = extract_embedding_from_cropped_face(img)

            if emb is None:
                continue

            pred_label, score, _ = predict_identity_from_embedding(
                emb, gallery_centroids, threshold=SIMILARITY_THRESHOLD
            )

            results.append({
                "race": row["race"],
                "gender": row["gender"],
                "age": row["age"],
                "pred_label": pred_label,
                "confidence": score
            })

        except Exception:
            continue

    return pd.DataFrame(results)
```

```python
bias_summary = fairface_results_renamed.groupby("ethnicity").agg({
    "confidence": ["mean", "count"]
})

bias_summary.columns = ["avg_confidence", "num_samples"]
bias_summary = bias_summary.sort_values("avg_confidence", ascending=False)

display(bias_summary)
```

```python
def simulate_video_from_frames(
    frames_dir,
    gallery_centroids,
    threshold=0.45,
    camera_id="CAM_SIM"
):
    frames_dir = Path(frames_dir)
    frame_paths = sorted(list(frames_dir.glob("*.jpg")))

    print(f"Processing {len(frame_paths)} frames...")

    for idx, img_path in enumerate(frame_paths):
        img = read_image(img_path)

        faces = detect_faces_and_embeddings(img)

        for face in faces:
            emb = l2_normalize(face.embedding.astype(np.float32))

            name, score, _ = predict_identity_from_embedding(
                emb, gallery_centroids, threshold
            )

            if name != "Unknown":
                append_attendance(
                    ATTENDANCE_CSV,
                    name,
                    camera_id,
                    score,
                    excel_path=ATTENDANCE_XLSX
                )

        print(f"Frame {idx+1} processed")

    print(" Video simulation complete")
```
